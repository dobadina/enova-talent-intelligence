```python
from google.colab import drive, userdata
import shutil, os

drive.mount('/content/drive')

token    = userdata.get('GITHUB_TOKEN')
email    = userdata.get('GITHUB_EMAIL')
username = userdata.get('GITHUB_USERNAME')
repo     = 'enova-talent-intelligence'
dst      = f'/content/{repo}'

os.system(f'git config --global user.email "{email}"')
os.system(f'git config --global user.name "{username}"')

# Clone fresh copy of the repo
os.system(f'rm -rf {dst}')
os.system(f'git clone https://{username}:{token}@github.com/{username}/{repo}.git {dst}')

# Copy Drive files into repo
src = f'/content/drive/MyDrive/{repo}'
shutil.copytree(src, dst, dirs_exist_ok=True)

# Commit and push
os.chdir(dst)
os.system('git add .')
os.system('git commit -m "Add bronze load, silver layer, gold layer"')
os.system(f'git push https://{username}:{token}@github.com/{username}/{repo}.git main')

print('Done.')
```
